## Initial Observations

* The dataset contains 7,262 records and 22 columns. The intended grain appears to be one row per customer, with `CustomerID` serving as the customer identifier.
* The dataset contains 5 numerical columns and 17 object/string columns. Of the string columns, `CustomerID` is an identifier rather than an analytical categorical variable, resulting in 16 categorical variables.
* `Age` ranges from 18 to 80 years, while `Tenure` ranges from 0 to 72 months.
* `MonthlyCharges` ranges from 18.00 to 101.20, and `TotalCharges` ranges from 0.35 to 7,084.30.
* The dataset contains missing values in `Age`, `TechSupport`, `PaymentMethod`, `MonthlyCharges`, and `TotalCharges`. The missing-value percentage ranges from approximately 0.80% to 1.40%.
* `CustomerID` contains 7,250 unique values across 7,262 records. Investigation showed that the repeated customer records are exact duplicate rows, so the duplicate copies can be addressed during the data-cleaning stage.
* In `Partner`, `Dependents`, `PhoneService`, and `PaperlessBilling`, inconsistent capitalization exists, such as `Yes`/`yes` and `No`/`NO`. These values represent the same business categories and should be standardized during data cleaning.
* `SeniorCitizen` is stored as an integer encoded with 0 and 1. Although its storage type is numerical, it should be treated analytically as a binary categorical variable.
* `Churn` contains two categories: `Yes` and `No`. There are 1,840 churned customers and 5,422 non-churned customers, giving an overall churn rate of approximately 25.34%.
* Some service-related columns contain the category `No internet service`. This represents a valid business condition and should not automatically be treated as a missing value.
* Further data cleaning and validation are required before performing detailed exploratory analysis.

## Conclusion

The initial data-understanding phase established the structure, grain, variable types, distributions, and major data-quality concerns in the customer dataset.

The dataset contains 7,262 customer records and 22 columns, with `Churn` representing the primary business outcome. The overall observed churn rate is approximately 25.34%.

Several data-quality issues were identified, including missing values, exact duplicate records, and inconsistent capitalization in selected categorical variables. `SeniorCitizen` is stored as a binary integer but will be treated as a categorical variable for analysis.

The initial numerical range checks did not identify obvious invalid values. Values such as `Tenure = 0` and low `TotalCharges` values will be retained for further investigation because they may represent legitimate customer situations rather than data errors.

The identified data-quality issues will be addressed systematically in the data-cleaning stage before conducting detailed exploratory analysis.
